```python
import matplotlib.pyplot as plt

# 1. Data
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [72, 74, 68, 79, 83, 85, 80]

# 2. Create the figure
plt.figure(figsize=(10, 5))

# 3. Plot the line with markers and custom styling
plt.plot(days, temperatures, marker='o', color='#2b5c8f', linewidth=2.5, markersize=8, label='Daily High')

# 4. Fill the area under the line
plt.fill_between(days, temperatures, color='#2b5c8f', alpha=0.15)

# 5. Add titles, labels, and grid
plt.title('Weekly Temperature Trend (Intermediate Level)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Days of the Week', fontsize=11)
plt.ylabel('Temperature (°F)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)

# Highlight the highest point automatically
max_temp = max(temperatures)
max_day = days[temperatures.index(max_temp)]
plt.annotate(f'Peak: {max_temp}°F', 
             xy=(max_day, max_temp), 
             xytext=(max_day, max_temp + 2),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6),
             ha='center', fontweight='bold')

# 6. Display
plt.tight_layout()
plt.show()
```
